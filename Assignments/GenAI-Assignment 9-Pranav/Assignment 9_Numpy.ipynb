## Task 2: Important Mathematical Operations

```python
A = np.array([10,20,30,40,50])
B = np.array([1,2,3,4,5])
print("Additon of A and B is:", A + B)
print("Subtraction of A and B is:", A - B)
print("Multiplication of A and B is:", A * B)
print("Division of A and B is:", (A / B).astype(int))
print("Square of A is:", A ** 2)
```

## Task 3: Important Nummpy Mathematical Formaulas

```python
values = np.array([2,4,6,8,10])
print("Square Root of element is:", np.sqrt(values))
print("Exponential of element is:", np.exp(values))
print("Logarithm of element is:", np.log(values))
print("Sum of all elements is:", np.sum(values))
print("Cumulative sum of all elements is:", np.cumsum(values))
```

## Task 4: Aggregation Operations

```python
data = np.array([[10,20,30],[40,50,60],[70,80,90]])
```

```python
print("Rowwise sum of elements in the array is:", np.sum(data, axis=1))
print("Columnwise sum of elements in the array is:", np.sum(data, axis=0))
print("Mininmum value in the array is:", np.min(data))
print("Maximum value in the array is:", np.max(data))
print("Mean of all elements in the array is:", np.mean(data))
```

## Task 5: Statistical Operations

```python
marks  = np.array([78,85,90,66,72,88,95,60])
```

```python
print("Mean of marks is:", np.mean(marks))
print("Median of marks is:", np.median(marks))
print("Variance of marks is:", np.var(marks))
print("Standard Deviation of marks is:", np.std(marks))
print("Minimum and Maximum marks are:", np.min(marks), "and", np.max(marks))
print("Range of the array is:", np.max(marks) - np.min(marks))
```

## Task 6: Percentiles and Sorting

```python
marks = np.sort(marks)
print("The 25th, 50th and 75th percentiles of the marks are:",np.percentile(marks, [25, 50, 75]))
avg_marks = np.mean(marks)
above_avg = marks[marks > avg_marks]
print("The students who scored above average marks are:", above_avg)
```

## Task 7: Mini Use Case - Sales Analysis

```python
sales = np.array([1200,1500,900,2000,1800,1700,1600])
```

```python
print("Total sales for the week is:", np.sum(sales))
print("Average sales for the week is:", np.mean(sales))
print("Highest and Lowest sales for the week are:", np.max(sales), "and", np.min(sales))
print("The standard deviation of sales for the week is:", np.std(sales))
avg_sales = np.mean(sales)
above_avg_sales = sales[sales > avg_sales]
print("The days with above average sales are:", above_avg_sales)
```
