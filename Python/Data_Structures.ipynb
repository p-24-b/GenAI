## List Methods 

```python
list_items = [1, 2, 3, 4, 5,7,8,9]
```

```python
# list_items.append(10)
# list_items.insert(4, 6)
# list_items.remove(10)   
##->removes first occurence of element
print(list_items)
```

## List Comprehension
Looping statements inside the list so that we can store elements inside the list in an iterable format also we have a conditional inside the list.


```python
even_squares = [x**2 for x in range(21) if x % 2 == 0]
```

```python
print(even_squares)
```
