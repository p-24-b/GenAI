# File Handling in Python


### Reading from a file

```python
filename = 'file_handling.txt'
file = open(filename, 'r')
```

```python
file_content = file.read()
print(file_content)
file.close()
```

### Writing into a file

```python
with open(filename, 'a') as file:
    file.write("\nThis is an additional line added to the file.")
```

```python
list = ['Hi this is Pranav Bajpai\n', 'I am a student of B.Tech in Metallurgical Engineering\n', 'I am currently learning Python programming language']
with open(filename, 'w') as file:
    file.writelines(list)
```

### Reading total lines, total words and total characters in the file

```python
def count_lines(filename):
    with open(filename, 'r') as file:
        lines = file.readlines()
        total_lines = len(lines)
        total_words = sum(len(line.split()) for line in lines)
        return total_lines, total_words
print("Number of lines in the file:", count_lines(filename))
```
