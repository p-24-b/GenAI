### Task 2: Ways of Reading a file

```python
with open('sales_data.txt', 'r') as file:
    data = file.read()
print(data)
```

```python
with open('sales_data.txt', 'r') as file:
    first_line = file.readline()
    print(first_line)
    file.seek(0)
    data = file.readlines()
    sales_data = [int(line.strip()) for line in data]
print(sales_data)
    
```

### Task 4: Generate Summary Report

```python
with open('sales_data.txt', 'r') as file:
    data = file.readlines()
    sales_data = [int(line.strip()) for line in data]
total_sales = sum(sales_data)
print("Total Sales:", total_sales)
highest_sale = max(sales_data)
print("Highest Sale:", highest_sale)
minimum_sale = min(sales_data)
print("Lowest Sale:", minimum_sale)
average_sale = total_sales / len(sales_data)
print("Average Sale:", average_sale)
```

### Task 5 : Create Product Info file

```python
product_dict = {}
for i in range(3):
    key = input("Enter product name: ")
    value = input("Enter product price: ")
    product_dict[key] = value

```

```python
with open ('product.txt','w') as file:
    file.write('Product|Price\n')
    file.writelines(i+'|'+product_dict[i]+'\n' for i in product_dict.keys())

```

```python
with open('product.txt','r') as file:
    data = file.read()
    print(data)
```

### Task 6: Read File Safely

```python
import os
filename = input("Enter the filename along with extension to be read: ")
if os.path.exists(filename):
    with open (filename,'r') as file:
        data = file.read()
        print(data)
else:
    print("File not found. Please check the filename")
```

### Task 7: Mini Project

```python
prices ={'Mouse':500,'Keyboard':800,'Monitor':7000,'Pendrive':400,'Camera':5000}
discount = int(input("Enter the discount percentage"))
filename = 'discount_report.txt'
with open (filename,'w') as file:
    file.write('Product| Original Price| Discounted Price \n')
    for i in prices.keys():
        discounted_price = prices[i]-(prices[i]*(discount/100))
        file.writelines(f'{i}|{prices[i]}|{discounted_price}\n')
```

```python
with open (filename,'r') as file:
    data = file.read()
    print(data)
```
