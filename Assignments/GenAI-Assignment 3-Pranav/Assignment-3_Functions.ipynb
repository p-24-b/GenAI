### Task 1 - Basic Function: Price After Discount

```python
def apply_discount(price, discount = 5):
    if discount > 60:
        raise ValueError("Discount cannot be greater than 60%")
    discounted_price = round((price - (price * discount / 100)),2)
    return discounted_price
```

```python
print(apply_discount(1000, 10))
print(apply_discount(500))
print(apply_discount(1000, 70))  # This will raise a ValueError
```

### Task 2 - Recursive Function


```python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    elif n == 0 or n == 1:
        return 1
    else:
       return n * factorial(n - 1)
```

```python
factorial(-3) # This will raise a ValueError
```

```python
factorial(5)
```

### Task 3 - Lambda Function: GST Calculator

```python
gst = lambda price: price + round(price * 0.18, 2)
print("The Goods and Service Tax on the price is:",gst(100))
```

### Task 4 - Map Functions

```python
prices = [100,250,400,1200,50]
```

```python
prices_with_gst = list(map(gst, prices))
print("Prices after adding GST:", prices_with_gst)
```

### Task 5 - Using filter: Filter expensive products

```python
prices = [100,250,400,1200,50,2000,850]
price_greater_than_500 = list(filter(lambda price: price > 500, prices))
price_less_than_equal_to_500 = list(filter(lambda price: price <= 500, prices))
print("Prices greater than 500:", price_greater_than_500)
print("Prices less than or equal to 500:", price_less_than_equal_to_500)
```

### Task 6 - Combined Utility Function

```python
def process_prices(prices):
    discounted_prices = list(map(lambda price:price - (price * 0.10), prices))
    filtered_prices = list(filter(lambda price: price > 300, discounted_prices))
    return discounted_prices,filtered_prices
```

```python
processed_prices = process_prices([100,500,900,50,750])
print(f"Processed prices after discount: {processed_prices[0]} and filtering: {processed_prices[1]}")
```

### Task 7 - Mini Problem 

```python
def add_price(price_list,price):
    price_list.append(price)
def get_average_price(price_list):
    if len(price_list) == 0:
        return 0
    return sum(price_list)/len(price_list)
def get_max_price(price_list):  
    if len(price_list) == 0:
        return 0
    return max(price_list)
```

```python
print("""Choose from the following options:
1. Add price
2. Show average price
3. Show highest price
q. Quit
""")
price_list = []
choice = input("Enter your choice: ")
while choice != 'q':
    if choice == '1':
        price = float(input("Enter the price to add: "))
        add_price(price_list, price)
        print(f"Price {price} added.")
    elif choice == '2':
        average_price = get_average_price(price_list)
        print(f"The average price is: {average_price}")
    elif choice == '3':
        max_price = get_max_price(price_list)
        print(f"The highest price is: {max_price}")
    else:
        print("Invalid choice. Please try again.")
    choice = input("Enter your choice: ")
```
